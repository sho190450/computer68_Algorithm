```python
Q = "Press = exit"
M = "Press = Main"
S = "Press = Start in Program"
state = True
while(state):
    print(Q)
    print(M)
    print(S)

    inp = input("Menu Selected :")
    if(inp == 'Q'):
        state = False
    elif(inp == 'M'):
        print("Home Menu")
    elif(inp == 'S'):
        runnnig_str = input("จำนวนรอบที่ต้องการวิ่ง : ")
        runnnig_int = int(runnnig_str)
        for i in range (runnnig_int):
            print("รอบที่ :", i+1)
    else:
        print("Home Menu")
```
